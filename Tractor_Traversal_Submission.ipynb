```python
### START FUNCTION
def tractor_movement_standard(field):
    path = []
    for r_idx, row in enumerate(field):
        for c_idx in range(len(row)):
            path.append((r_idx, c_idx))
    return path
### END FUNCTION

### START FUNCTION
def reverse_list(input_list):
    return input_list[::-1]
### END FUNCTION

### START FUNCTION
def tractor_movement_realistic(field, start_direction='right'):
    path = []
    for r_idx, row in enumerate(field):
        cols = list(range(len(row)))
        # Even index (0, 2, 4...) default to 'right' (no reverse), odd (1, 3, 5...) default to 'left' (reverse)
        # If start_direction is 'left', then even rows are reversed, and odd rows are normal.
        if start_direction == 'right':
            if r_idx % 2 == 1:
                cols = reverse_list(cols)
        elif start_direction == 'left':
            if r_idx % 2 == 0:
                cols = reverse_list(cols)

        for c_idx in cols:
            path.append((r_idx, c_idx))
    return path
### END FUNCTION

### START FUNCTION
def field_summary(field):
    num_rows = len(field)
    total_cells = sum(len(row) for row in field)
    if num_rows == 0:
        is_rectangular = True
    else:
        first_row_len = len(field[0])
        is_rectangular = all(len(row) == first_row_len for row in field)
    return num_rows, total_cells, is_rectangular
### END FUNCTION

### START FUNCTION
def total_fuel_used(*segment_costs):
    return sum(segment_costs)
### END FUNCTION

### START FUNCTION
def log_trip(tractor_id, **details):
    record = {'tractor_id': tractor_id}
    record.update(details)
    return record
### END FUNCTION
```
